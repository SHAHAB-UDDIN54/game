```python
print("=" * 60)
print("BAKESUITE AI - DEMAND FORECASTING")
print("=" * 60)

print("\nDataset Shape:")
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")

print("\nColumn Names:")
for i, column in enumerate(df.columns, 1):
    print(f"{i}. {column}")

# ==========================================
# 3. DATA TYPES
# ==========================================

print("\nData Types:")
print(df.dtypes)

# ==========================================
# 4. MISSING VALUES
# ==========================================

print("\nMissing Values:")
missing = df.isnull().sum()
display(missing[missing > 0])

print(f"Total Missing Values: {df.isnull().sum().sum()}")

# ==========================================
# 5. DUPLICATES
# ==========================================

duplicates = df.duplicated().sum()

print(f"\nDuplicate Rows: {duplicates}")

# ==========================================
# 6. DATA PREVIEW
# ==========================================

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("=" * 60)
print("INITIAL DATASET INSPECTION COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# STEP 6 - DUPLICATE ANALYSIS
# ============================================================

print("=" * 60)
print("DUPLICATE ANALYSIS")
print("=" * 60)

print("\nTotal rows:")
print(len(df))

print("\nExact duplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate examples:")
duplicates = df[df.duplicated(keep=False)].sort_values(
    ["TransactionNo", "Items", "DateTime"]
)

display(duplicates.head(30))

print("\nNumber of duplicate TransactionNo + Items combinations:")

transaction_item_counts = (
    df.groupby(["TransactionNo", "Items"])
      .size()
      .reset_index(name="Count")
)

display(
    transaction_item_counts[
        transaction_item_counts["Count"] > 1
    ].head(30)
)

print("=" * 60)
```
